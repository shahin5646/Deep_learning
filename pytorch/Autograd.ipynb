```python
x = torch.tensor(9.0 ,requires_grad=True)

with torch.no_grad():
    y = x**2 + 55

print(y.requires_grad)
```

```python
x = torch.tensor(9.0 ,requires_grad=True)
y = x**2 + 24

grad = torch.autograd.grad(y,x)
print(grad)
```

### Creating a Simple forward Backword

```python
x = torch.tensor(4.0 , requires_grad=True)
y = x**2 # Forward Pass
z = torch.sin(y)
print(f"Value of Z: {z.item()}")
# backward Pass
z.backward() 
print(f"Gradient of z with respect to x: {x.grad.item()}")
```

```python
import torch.nn as nn

# Input and target
x = torch.tensor([1.0, 2.0], requires_grad=True)  # Input
y = torch.tensor([5.0])  # Target

# Weights and biases
W1 = torch.tensor([[0.5, -0.2], [0.3, 0.8]], requires_grad=True)  # Input to hidden
b1 = torch.tensor([0.1, -0.1], requires_grad=True)  # Bias for hidden layer
W2 = torch.tensor([[0.7], [-0.5]], requires_grad=True)  # Hidden to output
b2 = torch.tensor([0.3], requires_grad=True)  # Bias for output layer

# Forward pass
z1 = torch.matmul(x, W1) + b1  # Input to hidden layer
h = torch.relu(z1)  # Activation function
z2 = torch.matmul(h, W2) + b2  # Hidden to output layer
y_pred = z2  # Output (no activation)

# Loss
loss = 0.5 * (y_pred - y) ** 2  # MSE Loss

# Backward pass
loss.backward()

# Gradients
print("Gradients:")
print(f"dL/dW1: {W1.grad}")
print(f"dL/db1: {b1.grad}")
print(f"dL/dW2: {W2.grad}")
print(f"dL/db2: {b2.grad}")
```
