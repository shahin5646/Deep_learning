```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        print(f"{self.name} is barking........")

    def info(self):
        print(f"{self.name} is a {self.breed}")


german_shepherd = Dog("Rex", "German Shepherd")
german_shepherd.bark()
german_shepherd.info()
```

**torch.manual_seed(100) not completely random**
- they're generated by a formula starting from a "seed."
- Same seed same sequence of numbers, every time, on any machine.

- **Why do we need this? When you're debugging a model, you want the same random weights every run, so you can tell if a change you made actually fixed something — instead of wondering "did it improve, or did I just get luckier random numbers this time?"**

```python
torch.manual_seed(100)
torch.rand(2,3)
torch.rand(2,3)
```

##  What `_like` Functions Actually Copy

`empty_like(x)`, `zeros_like(x)`, and `ones_like(x)` copy two things from `x` by default:

- **Shape** → `(2, 3)`
- **Data Type (`dtype`)** → `torch.int64`

```python
torch.empty_like(x)

```

```python
torch.zeros_like(x)
```

```python
torch.ones_like(x)
```

```python
torch.rand_like(x, dtype=torch.float32)
```

```python
x + 2      # adds 2 to every element
x - 2      # subtracts 2 from every element
x * 3      # multiplies every element by 3
x / 3      # divides every element by 3
(x * 100) // 3     # integer division (drops remainder) after scaling up
((x * 100) // 3) % 2   # modulo — remainder after dividing by 2
x ** 2     # squares every element
```
